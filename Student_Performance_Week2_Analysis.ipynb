```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/StudentsPerformance.csv")

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)
print(df.head())
```

```python
# Visualization 1: Average Scores by Gender

average_scores = df.groupby("gender")[["math score", "reading score", "writing score"]].mean()

print(average_scores)

average_scores.plot(kind="bar", figsize=(8, 5))

plt.title("Average Student Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Distribution of Student Scores

plt.figure(figsize=(10, 6))

sns.histplot(df["math score"], kde=True, label="Math Score", alpha=0.5)
sns.histplot(df["reading score"], kde=True, label="Reading Score", alpha=0.5)
sns.histplot(df["writing score"], kde=True, label="Writing Score", alpha=0.5)

plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Visualization 3: Correlation Heatmap

plt.figure(figsize=(8, 6))

correlation = df[["math score", "reading score", "writing score"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=0,
    vmax=1,
    linewidths=0.5
)

plt.title("Correlation Between Student Scores")
plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Test Preparation and Student Performance

prep_scores = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()

print(prep_scores)

prep_scores.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.title("Impact of Test Preparation on Student Performance")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python
# 5. Student Performance by Parental Education Level

plt.figure(figsize=(10,6))

sns.barplot(
    data=df,
    x='parental level of education',
    y='math score',
    errorbar=None
)

plt.title('Math Score by Parental Level of Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Average Math Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='lunch',
    y='math score'
)

plt.title('Math Score Distribution by Lunch Type')
plt.xlabel('Lunch Type')
plt.ylabel('Math Score')

plt.tight_layout()
plt.show()
```
