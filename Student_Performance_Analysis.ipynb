```python
import matplotlib.pyplot as plt

df[['math score', 'reading score', 'writing score']].hist(
    figsize=(10, 4),
    bins=10
)

plt.suptitle("Distribution of Student Scores")
plt.show()
```

```python
df.groupby('gender')[['math score', 'reading score', 'writing score']].mean().plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title("Average Scores by Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

score_corr = df[['math score', 'reading score', 'writing score']].corr()

sns.heatmap(score_corr, annot=True, cmap='Blues')

plt.title("Correlation Between Student Scores")
plt.show()
```
